\Missing Value Analysis

```python
def missing_report(df,table_name):
    miss_count= df.isna().sum()
    miss_pct=((miss_count/len(df))*100).round(2)
    report=pd.DataFrame({
            "table":table_name,
            "columns":df.columns,
            "missing_count":miss_count,
            "missing_pct":miss_pct
    })
    report=report[report["missing_count"] >0].sort_values("missing_pct",ascending=False).reset_index(drop=True)
    return report
```

```python
master_missing=missing_report(master,"master")
print("===== MASTER MISSING VALUES =====")
print(master_missing.to_string(index=False) if len(master_missing) else "No missing!")
```

```python
items_missing=missing_report(items_enriched,"items_enriched")
print("\n===== ITEMS_ENRICHED MISSING =====")
print(items_missing.to_string(index=False) if len(items_missing) else "No missing!")
```

\Consistency Checks

```python
def check_categoricals(df, table_name, expected_values=None, max_unique=20):
   
    print(f"\n{'='*60}\n  CATEGORICAL CONSISTENCY: {table_name}\n{'='*60}")
    
    cat_cols = []
    for c in df.select_dtypes(include="object").columns:
        nu = df[c].nunique()
        if nu <= max_unique and 'id' not in c.lower() \
           and 'date' not in c and 'timestamp' not in c and 'name' not in c.lower():
            cat_cols.append(c)
    
    for col in cat_cols:
        vc = df[col].value_counts(dropna=False)
        print(f"\n {col}  ({df[col].nunique()} unique values):")
        for val, cnt in vc.items():
            pct = cnt / len(df) * 100
            print(f"    {str(val):20s}: {cnt:>6,} ({pct:5.1f}%)")
        
        if expected_values and col in expected_values:
            actual = set(df[col].dropna().unique())
            expected = set(expected_values[col])
            unexpected = actual - expected
            if unexpected:
                print(f"     UNEXPECTED values found: {unexpected}")
            else:
                print(f"     All values valid")
```
